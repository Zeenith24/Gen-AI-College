```python
import pandas as pd
from transformers import pipeline
```

```python
file_path = "/content/IMDB-Dataset - IMDB-Dataset.csv"
```

```python
df = pd.read_csv(file_path, engine='python', on_bad_lines='skip')
print("Dataset loaded successfully!")
print("Number of reviews:", len(df))
print(df.head())
```

```python
sentiment_analyzer = pipeline("sent"analysis",
  model"ilbert-base-"d-finetuned-sst-2-english"
)
```

```python
for i in range(5):
  review = df.iloc[i]["review"]
  prediction = sentiment_analyzer(review[:512])[0]
  print(" \nReview ", i + 1)
  print(" Actual Sentiment   : ", df.iloc[i]["sentiment"].upper())
  print(" Predicted Sentiment: ", prediction["label"])
  print(" Confidence          : ", round(prediction["score"], 4))
  print(" Review              : ", review[:100], " ... ")
```
